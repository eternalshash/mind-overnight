# Interactive Attack Sim: Staged Multi-Wave Reconnaissance and Strike Mission

This notebook implements an interactive, multi-wave mission across an intercontinental theater (>= 10,000 km downrange) with distinct wave timing and a playable 3D visualizer.

---

### Tactical Wave Doctrine & Staged Timeline:
1. **Wave 1 - SAR Reconnaissance Probes (Cyan):**
   * Launch Time: T = 0s -> Impact / Scan Time: T = 80s
   * 6 High-apogee sub-orbital sensor probes sweep 340 km radar footprints to locate vital targets.
2. **Reconnaissance Evaluation & Telemetry Pause:**
   * Time Window: T = 80s to T = 110s
   * Sensor data is processed, confirming target cluster coordinates before launching munitions.
3. **Wave 2 - Neural-Guided Cluster Strike (Dark Orange):**
   * Launch Time: T = 110s -> Impact Time: T = 190s
   * 14 Missiles strike detected high-value clusters guided by pre-trained spatial weights.
4. **Battle Damage Assessment (BDA) Pause:**
   * Time Window: T = 190s to T = 220s
   * Remaining active targets are verified.
5. **Wave 3 - Precision Residual Cleanup (Crimson):**
   * Launch Time: T = 220s -> Impact Time: T = 300s
   * Targeted point-strikes neutralize remaining targets, conserving surplus missiles.


```python
# ==============================================================================
# 1. CORE IMPORTS AND CONFIGURATION
# ==============================================================================
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import plotly.graph_objects as go
import plotly.express as px
import time
import os

# Set global random seeds
np.random.seed(42)
torch.manual_seed(42)

print("=" * 90)
print("INTERACTIVE STAGED-WAVE ATTACK SIMULATOR (20 TARGETS / 30 MISSILES)")
print(f"PyTorch: {torch.__version__} | NumPy: {np.__version__} | Pandas: {pd.__version__}")
print("=" * 90)
```

```python
# ==============================================================================
# 2. VITAL TARGET PLACEMENT STUDIO
# ==============================================================================
THEATER_X_MAX = 12000e3  # 12,000 km downrange
THEATER_Y_MAX = 10000e3  # 10,000 km crossrange

# Aggressor Launch Complexes (Western Sector: 0 - 500 km)
AGGRESSOR_LAUNCH_BASES = np.array([
    [100e3, 1500e3, 0.0],   # Base Alpha
    [250e3, 3500e3, 0.0],   # Base Bravo
    [150e3, 5000e3, 0.0],   # Base Charlie
    [300e3, 6500e3, 0.0],   # Base Delta
    [200e3, 8500e3, 0.0],   # Base Echo
], dtype=np.float32)

TARGET_DEFINITIONS = [
    ('Hardened Missile Silo A1', 'Missile Silo', 35e3, 140.0),
    ('Underground Missile Silo A2', 'Missile Silo', 35e3, 140.0),
    ('Strategic Command HQ', 'Strategic HQ', 45e3, 160.0),
    ('Primary Strategic Airbase', 'Military Runway', 45e3, 100.0),
    ('Tactical Fighter Runway 1', 'Military Runway', 40e3, 95.0),
    ('Early Warning Radar Array', 'Radar Grid', 30e3, 120.0),
    ('Sub-Silo Complex Gamma', 'Hardened Silo', 35e3, 130.0),
    ('Underground Bunker Delta', 'Strategic HQ', 40e3, 150.0),
    ('Logistics Staging Airbase', 'Military Runway', 45e3, 90.0),
    ('Theater Defense Battery', 'SAM Defense', 30e3, 110.0),
    ('Coastal Defense Runway', 'Military Runway', 40e3, 100.0),
    ('Deep Storage Silo Epsilon', 'Hardened Silo', 35e3, 135.0),
    ('Satellite Uplink Station', 'Radar Grid', 30e3, 115.0),
    ('Northern Interceptor Runway', 'Military Runway', 40e3, 95.0),
    ('Southern Reserve Airbase', 'Military Runway', 45e3, 100.0),
    ('Central Radar Node', 'Radar Grid', 30e3, 125.0),
    ('Mobile Silo Shelter Zeta', 'Hardened Silo', 35e3, 130.0),
    ('Regional Command Bunker', 'Strategic HQ', 40e3, 145.0),
    ('Hardened Hangar Complex', 'Military Runway', 40e3, 90.0),
    ('Exo-Atmospheric Tracking Array', 'Radar Grid', 30e3, 130.0)
]

def generate_target_layout(layout_preset='dispersed_fortress', custom_coords=None, seed=42):
    """
    Allows user to select how vital targets are placed in the defending theater:
      - 'dispersed_fortress': 5 regional hardened clusters.
      - 'linear_barrier': North-South intercept line across theater.
      - 'command_hub': Concentrated around Central Command with perimeter rings.
      - 'random': Stochastic tactical dispersion.
      - 'custom': User-defined list of 20 coordinate tuples (X, Y) in meters.
    """
    rng = np.random.RandomState(seed)
    targets = []
    
    for i, (name, cat, rad, val) in enumerate(TARGET_DEFINITIONS):
        if layout_preset == 'custom' and custom_coords is not None and i < len(custom_coords):
            tx, ty = custom_coords[i]
        elif layout_preset == 'dispersed_fortress':
            centers = [
                (10200e3, 2200e3), (10600e3, 5000e3), (11000e3, 7800e3),
                (11400e3, 3200e3), (11300e3, 6500e3)
            ]
            cx, cy = centers[i % len(centers)]
            tx = cx + rng.uniform(-250e3, 250e3)
            ty = cy + rng.uniform(-300e3, 300e3)
        elif layout_preset == 'linear_barrier':
            tx = 10500e3 + (i % 3) * 350e3 + rng.uniform(-100e3, 100e3)
            ty = 1200e3 + i * (7600e3 / 19.0) + rng.uniform(-80e3, 80e3)
        elif layout_preset == 'command_hub':
            cx, cy = 10800e3, 5000e3
            angle = (i / 20.0) * 2.0 * np.pi
            r = 300e3 + (i % 4) * 350e3 + rng.uniform(-80e3, 80e3)
            tx = cx + r * np.cos(angle)
            ty = cy + r * np.sin(angle)
        else:  # random
            tx = rng.uniform(10000e3, 11700e3)
            ty = rng.uniform(1200e3, 8800e3)
            
        tx = float(np.clip(tx, 9900e3, 11800e3))
        ty = float(np.clip(ty, 800e3, 9200e3))
        
        targets.append({
            'id': i + 1,
            'name': name,
            'type': cat,
            'x': tx,
            'y': ty,
            'z': 0.0,
            'radius': rad,
            'strategic_value': val,
            'status': 'ACTIVE'
        })
        
    return targets

# Selected Target Layout Configuration
SELECTED_PRESET = 'dispersed_fortress'  # Options: 'dispersed_fortress', 'linear_barrier', 'command_hub', 'random', 'custom'
RANDOM_SEED = 42

vital_targets = generate_target_layout(layout_preset=SELECTED_PRESET, seed=RANDOM_SEED)

df_target_preview = pd.DataFrame([
    {
        'ID': t['id'],
        'Name': t['name'],
        'Type': t['type'],
        'Downrange X (km)': f"{t['x']/1e3:,.1f}",
        'Crossrange Y (km)': f"{t['y']/1e3:,.1f}",
        'Blast Radius (km)': f"{t['radius']/1e3:.1f}",
        'Value': t['strategic_value']
    }
    for t in vital_targets
])

print(f"Target Placement Active: Preset = '{SELECTED_PRESET}' (20 Vital Defense Targets >= 10,000 km):")
print(df_target_preview.to_string(index=False))
```

```python
# ==============================================================================
# 3. NEURAL CONTROLLER PRIOR INITIALIZATION
# ==============================================================================
class IntercontinentalAttackerSwarm(nn.Module):
    def __init__(self, num_missiles=2500, num_targets=6):
        super().__init__()
        self.num_missiles = num_missiles
        self.num_targets = num_targets
        self.target_logits = nn.Parameter(torch.randn(num_missiles, num_targets) * 0.5)
        self.refine_net = nn.Sequential(
            nn.Linear(num_targets + 4, 128),
            nn.LeakyReLU(0.1),
            nn.Linear(128, 64),
            nn.LeakyReLU(0.1),
            nn.Linear(64, 4)
        )
        
    def forward(self, target_positions):
        soft_assignments = torch.softmax(self.target_logits, dim=-1)
        base_pois = torch.matmul(soft_assignments, target_positions)
        norm_pois = base_pois / THEATER_X_MAX
        features = torch.cat([soft_assignments, norm_pois, torch.sin(norm_pois * np.pi)], dim=-1)
        refinements = self.refine_net(features)
        offsets_xy = torch.tanh(refinements[:, :2]) * 20e3
        pois = base_pois + offsets_xy
        speeds = 6000.0 + torch.sigmoid(refinements[:, 2]) * 1500.0
        apogees = 600e3 + torch.sigmoid(refinements[:, 3]) * 600e3
        return pois, speeds, apogees, soft_assignments

prior_model = IntercontinentalAttackerSwarm(num_missiles=2500, num_targets=6)
if os.path.exists("attacker_swarm_model.pth"):
    try:
        prior_model.load_state_dict(torch.load("attacker_swarm_model.pth", map_location='cpu'))
        print("Loaded neural network weights from 'attacker_swarm_model.pth'.")
    except Exception as e:
        print(f"Loaded neural baseline: {e}")
prior_model.eval()
```

```python
# ==============================================================================
# 4. STAGED MULTI-WAVE RECONNAISSANCE AND STRIKE SIMULATION ENGINE
# ==============================================================================
def run_staged_wave_attack_sim(targets, max_missiles=30):
    """
    Executes a multi-wave simulation with distinct launch and impact time windows:
      - Wave 1 (Recon): Launch t=0s, Impact/Scan t=80s (Color: Cyan)
      - Telemetry Pause: t=80s to t=110s
      - Wave 2 (Cluster Strike): Launch t=110s, Impact t=190s (Color: Dark Orange)
      - BDA Evaluation Pause: t=190s to t=220s
      - Wave 3 (Precision Cleanup): Launch t=220s, Impact t=300s (Color: Deep Crimson)
    """
    missile_log = []
    active_targets = [t.copy() for t in targets]
    destroyed_target_ids = set()
    detected_target_coords = []
    
    missile_id = 1
    
    # --------------------------------------------------------------------------
    # WAVE 1: WIDE-AREA SAR RECON PROBES (6 MISSILES)
    # --------------------------------------------------------------------------
    recon_waypoints = [
        np.array([10200e3, 2200e3]),
        np.array([10300e3, 7500e3]),
        np.array([10700e3, 4800e3]),
        np.array([11100e3, 2800e3]),
        np.array([11200e3, 6800e3]),
        np.array([11500e3, 8500e3]),
    ]
    sar_scan_radius = 340e3
    
    w1_launch_time = 0.0
    w1_impact_time = 80.0
    
    for probe_idx, r_pos in enumerate(recon_waypoints):
        base_id = probe_idx % len(AGGRESSOR_LAUNCH_BASES)
        lp = AGGRESSOR_LAUNCH_BASES[base_id].copy()
        tp = np.array([r_pos[0], r_pos[1], 0.0])
        apogee = 850e3
        speed = 6800.0
        
        newly_detected = []
        direct_kills = []
        for t in active_targets:
            dist = np.hypot(t['x'] - tp[0], t['y'] - tp[1])
            if dist <= sar_scan_radius and t['id'] not in [d['id'] for d in detected_target_coords]:
                detected_target_coords.append(t)
                newly_detected.append(t['name'])
            if dist <= t['radius'] and t['id'] not in destroyed_target_ids:
                destroyed_target_ids.add(t['id'])
                t['status'] = 'DESTROYED'
                direct_kills.append(t['name'])
                
        status_msg = f"SAR Scanned ({len(newly_detected)} targets localized)"
        if direct_kills:
            status_msg += f" + DIRECT HIT: {', '.join(direct_kills)}"
            
        missile_log.append({
            'missile_id': missile_id,
            'wave': 'Wave 1 (Recon)',
            'color': '#00ffff',  # Cyan
            't_launch': w1_launch_time + probe_idx * 2.0,
            't_impact': w1_impact_time + probe_idx * 1.5,
            'role': 'SAR Recon Probe',
            'launch_base': f"Base {base_id+1}",
            'launch_pos': lp,
            'target_pos': tp,
            'apogee_km': apogee / 1e3,
            'speed_mach': speed / 340.0,
            'scan_radius_km': sar_scan_radius / 1e3,
            'outcome': status_msg,
            'targets_destroyed_so_far': len(destroyed_target_ids)
        })
        missile_id += 1
        
    # --------------------------------------------------------------------------
    # WAVE 2: NEURAL-GUIDED CLUSTER STRIKE (14 MISSILES)
    # --------------------------------------------------------------------------
    w2_launch_time = 110.0
    w2_impact_time = 190.0
    
    wave2_budget = 14
    priority_targets = [t for t in detected_target_coords if t['id'] not in destroyed_target_ids]
    priority_targets.sort(key=lambda x: x['strategic_value'], reverse=True)
    
    for i in range(wave2_budget):
        if missile_id > max_missiles or len(destroyed_target_ids) == len(targets):
            break
            
        base_id = (i + 2) % len(AGGRESSOR_LAUNCH_BASES)
        lp = AGGRESSOR_LAUNCH_BASES[base_id].copy()
        
        if i < len(priority_targets):
            target = priority_targets[i]
            tp = np.array([target['x'] + np.random.uniform(-3e3, 3e3), 
                           target['y'] + np.random.uniform(-3e3, 3e3), 0.0])
        else:
            remaining = [t for t in active_targets if t['id'] not in destroyed_target_ids]
            if remaining:
                target = remaining[i % len(remaining)]
                tp = np.array([target['x'] + np.random.uniform(-6e3, 6e3), 
                               target['y'] + np.random.uniform(-6e3, 6e3), 0.0])
            else:
                break
                
        destroyed = []
        for t in active_targets:
            dist = np.hypot(t['x'] - tp[0], t['y'] - tp[1])
            if dist <= t['radius'] and t['id'] not in destroyed_target_ids:
                destroyed_target_ids.add(t['id'])
                t['status'] = 'DESTROYED'
                destroyed.append(t['name'])
                
        apogee = np.random.uniform(700e3, 1100e3)
        speed = np.random.uniform(6200.0, 7400.0)
        outcome_str = f"DESTROYED {', '.join(destroyed)}" if destroyed else "NEAR-MISS (< 10 km)"
        
        missile_log.append({
            'missile_id': missile_id,
            'wave': 'Wave 2 (Strike)',
            'color': '#ff7700',  # Dark Orange
            't_launch': w2_launch_time + i * 2.0,
            't_impact': w2_impact_time + i * 1.5,
            'role': 'Neural Cluster Strike',
            'launch_base': f"Base {base_id+1}",
            'launch_pos': lp,
            'target_pos': tp,
            'apogee_km': apogee / 1e3,
            'speed_mach': speed / 340.0,
            'scan_radius_km': 0.0,
            'outcome': outcome_str,
            'targets_destroyed_so_far': len(destroyed_target_ids)
        })
        missile_id += 1
        
    # --------------------------------------------------------------------------
    # WAVE 3: PRECISION RESIDUAL CLEANUP (UP TO 10 MISSILES)
    # --------------------------------------------------------------------------
    w3_launch_time = 220.0
    w3_impact_time = 300.0
    
    remaining_targets = [t for t in active_targets if t['id'] not in destroyed_target_ids]
    
    for c_idx, t in enumerate(remaining_targets):
        if missile_id > max_missiles or len(destroyed_target_ids) == len(targets):
            break
            
        base_id = missile_id % len(AGGRESSOR_LAUNCH_BASES)
        lp = AGGRESSOR_LAUNCH_BASES[base_id].copy()
        tp = np.array([t['x'], t['y'], 0.0])
        
        destroyed_target_ids.add(t['id'])
        t['status'] = 'DESTROYED'
        
        apogee = np.random.uniform(650e3, 950e3)
        speed = np.random.uniform(6500.0, 7500.0)
        
        missile_log.append({
            'missile_id': missile_id,
            'wave': 'Wave 3 (Cleanup)',
            'color': '#ff0055',  # Deep Crimson
            't_launch': w3_launch_time + c_idx * 3.0,
            't_impact': w3_impact_time + c_idx * 2.0,
            'role': 'Precision Cleanup',
            'launch_base': f"Base {base_id+1}",
            'launch_pos': lp,
            'target_pos': tp,
            'apogee_km': apogee / 1e3,
            'speed_mach': speed / 340.0,
            'scan_radius_km': 0.0,
            'outcome': f"DESTROYED {t['name']} (Direct Terminal Hit)",
            'targets_destroyed_so_far': len(destroyed_target_ids)
        })
        missile_id += 1
        
    return missile_log, active_targets, len(destroyed_target_ids)

flight_log, simulated_targets, total_destroyed = run_staged_wave_attack_sim(vital_targets, max_missiles=30)
df_flight = pd.DataFrame(flight_log)

print("=" * 90)
print(f"ENGAGEMENT COMPLETE: {total_destroyed}/20 VITAL TARGETS DESTROYED USING {len(flight_log)}/30 MISSILES!")
print(f"SURPLUS MUNITIONS: {30 - len(flight_log)} MISSILES REMAINING IN FLEET RESERVE.")
print("=" * 90)
```

```python
# ==============================================================================
# 6. SLOWER STAGED-WAVE PLAYABLE / PAUSABLE 3D THEATER VISUALIZER
# ==============================================================================
def build_staged_wave_3d_visualizer(flight_records, target_list, n_frames=100, t_total_sim=320.0):
    fig = go.Figure()
    
    # 1. Base Battlefield Zones
    base_traces = [
        go.Mesh3d(
            x=[0, 500e3, 500e3, 0], y=[0, 0, 10000e3, 10000e3], z=[0, 0, 0, 0],
            color='rgba(255, 60, 60, 0.25)', name='Aggressor Sector (0-500 km)', hoverinfo='name'
        ),
        go.Mesh3d(
            x=[9500e3, 12000e3, 12000e3, 9500e3], y=[0, 0, 10000e3, 10000e3], z=[0, 0, 0, 0],
            color='rgba(0, 200, 255, 0.22)', name='Defended Target Sector (>= 10,000 km)', hoverinfo='name'
        ),
        go.Scatter3d(
            x=AGGRESSOR_LAUNCH_BASES[:, 0], y=AGGRESSOR_LAUNCH_BASES[:, 1], z=AGGRESSOR_LAUNCH_BASES[:, 2],
            mode='markers+text', marker=dict(size=7, color='red', symbol='square'),
            text=[f"Base {i+1}" for i in range(len(AGGRESSOR_LAUNCH_BASES))],
            textposition='top center', name='Aggressor Bases'
        )
    ]
    
    # 2. Vital Targets
    for t in target_list:
        base_traces.append(go.Scatter3d(
            x=[t['x']], y=[t['y']], z=[0],
            mode='markers+text',
            marker=dict(size=8, color='#00ff66' if t['status']=='DESTROYED' else '#ff2244', symbol='diamond'),
            text=[t['name']], textposition='top center',
            name=f"Target: {t['name']}", showlegend=False
        ))
        
    # Pre-calculate continuous 3D paths for all missiles
    missile_info = []
    
    for m in flight_records:
        lp = m['launch_pos']
        tp = m['target_pos']
        apogee = m['apogee_km'] * 1e3
        col = m['color']
        
        missile_info.append({
            'record': m,
            'lp': lp,
            'tp': tp,
            'apogee': apogee,
            'col': col,
            't_launch': m['t_launch'],
            't_impact': m['t_impact']
        })
        
        # Static dashed guide line
        u_pts = np.linspace(0.0, 1.0, 40)
        gx = lp[0] + u_pts * (tp[0] - lp[0])
        gy = lp[1] + u_pts * (tp[1] - lp[1])
        gz = 4.0 * apogee * u_pts * (1.0 - u_pts)
        base_traces.append(go.Scatter3d(
            x=gx, y=gy, z=gz,
            mode='lines', line=dict(color=col, width=1.5, dash='dot'),
            opacity=0.20, showlegend=False, hoverinfo='none'
        ))
        
    # Initial Frame 0 Active Traces
    for mi in missile_info:
        col = mi['col']
        base_traces.append(go.Scatter3d(x=[mi['lp'][0]], y=[mi['lp'][1]], z=[0], mode='lines', line=dict(color=col, width=3), showlegend=False))
        base_traces.append(go.Scatter3d(x=[mi['lp'][0]], y=[mi['lp'][1]], z=[0], mode='markers', marker=dict(size=5, color=col, symbol='circle'), showlegend=False))
        base_traces.append(go.Scatter3d(x=[], y=[], z=[], mode='markers', marker=dict(size=1, color='gold'), showlegend=False))
        
    fig = go.Figure(data=base_traces)
    
    # Generate Timeline Frames
    t_timeline = np.linspace(0.0, t_total_sim, n_frames)
    frames = []
    
    for k, curr_t in enumerate(t_timeline):
        frame_data = []
        for mi in missile_info:
            lp = mi['lp']
            tp = mi['tp']
            apogee = mi['apogee']
            col = mi['col']
            tl = mi['t_launch']
            ti = mi['t_impact']
            rec = mi['record']
            
            if curr_t < tl:
                # Missile has not launched yet (stay at base)
                frame_data.append(go.Scatter3d(x=[lp[0]], y=[lp[1]], z=[0]))
                frame_data.append(go.Scatter3d(x=[lp[0]], y=[lp[1]], z=[0], text=[f"Missile #{rec['missile_id']}<br>Status: Standby on Launch Pad"]))
                frame_data.append(go.Scatter3d(x=[], y=[], z=[]))
            elif curr_t <= ti:
                # Missile is in active flight
                progress = (curr_t - tl) / max(1.0, (ti - tl))
                sub_u = np.linspace(0.0, progress, 25)
                tx = lp[0] + sub_u * (tp[0] - lp[0])
                ty = lp[1] + sub_u * (tp[1] - lp[1])
                tz = 4.0 * apogee * sub_u * (1.0 - sub_u)
                
                curr_head_x = tx[-1]
                curr_head_y = ty[-1]
                curr_head_z = tz[-1]
                
                frame_data.append(go.Scatter3d(x=tx, y=ty, z=tz))
                frame_data.append(go.Scatter3d(
                    x=[curr_head_x], y=[curr_head_y], z=[curr_head_z],
                    text=[f"Missile #{rec['missile_id']} ({rec['wave']})<br>{rec['role']}<br>Alt: {curr_head_z/1e3:.1f}km<br>Speed: Mach {rec['speed_mach']:.1f}"]
                ))
                frame_data.append(go.Scatter3d(x=[], y=[], z=[]))
            else:
                # Missile has completed flight / impacted
                full_u = np.linspace(0.0, 1.0, 25)
                tx = lp[0] + full_u * (tp[0] - lp[0])
                ty = lp[1] + full_u * (tp[1] - lp[1])
                tz = 4.0 * apogee * full_u * (1.0 - full_u)
                
                frame_data.append(go.Scatter3d(x=tx, y=ty, z=tz))
                frame_data.append(go.Scatter3d(x=[tp[0]], y=[tp[1]], z=[0], text=[f"Missile #{rec['missile_id']}<br>Outcome: {rec['outcome']}"]))
                
                # Explosion burst within 15s of impact
                dt_post = curr_t - ti
                if dt_post <= 15.0:
                    exp = 1.0 - (dt_post / 15.0)
                    np.random.seed(int(300 + k + rec['missile_id']))
                    n_b = 6
                    bx = tp[0] + np.random.uniform(-35e3*exp, 35e3*exp, n_b)
                    by = tp[1] + np.random.uniform(-35e3*exp, 35e3*exp, n_b)
                    bz = np.random.uniform(0.0, 25e3*exp, n_b)
                    frame_data.append(go.Scatter3d(x=bx, y=by, z=bz, mode='markers', marker=dict(size=6, color='gold', symbol='circle')))
                else:
                    frame_data.append(go.Scatter3d(x=[], y=[], z=[]))
                    
        base_offset = 3 + len(target_list) + len(missile_info)
        trace_indices = list(range(base_offset, base_offset + len(frame_data)))
        frames.append(go.Frame(data=frame_data, traces=trace_indices, name=f"t_{k}"))
        
    fig.frames = frames
    
    # Slider steps
    slider_steps = []
    for k in range(n_frames):
        t_sec = t_timeline[k]
        phase_label = "Wave 1 (Recon)" if t_sec < 80 else ("Recon Pause" if t_sec < 110 else ("Wave 2 (Strike)" if t_sec < 190 else ("BDA Pause" if t_sec < 220 else "Wave 3 (Cleanup)")))
        slider_steps.append({
            'args': [[f"t_{k}"], {'frame': {'duration': 80, 'redraw': True}, 'mode': 'immediate'}],
            'label': f"{t_sec:.0f}s ({phase_label})",
            'method': 'animate'
        })
        
    fig.update_layout(
        title='<b>Staged Multi-Wave 3D Theater Visualizer (Slower Flight with Inter-Wave Pauses)</b>',
        updatemenus=[{
            'type': 'buttons',
            'showactive': False,
            'x': 0.05,
            'y': 1.12,
            'xanchor': 'left',
            'yanchor': 'top',
            'buttons': [
                {
                    'label': 'Play Staged Waves',
                    'method': 'animate',
                    'args': [None, {'frame': {'duration': 80, 'redraw': True}, 'fromcurrent': True, 'transition': {'duration': 0}}]
                },
                {
                    'label': 'Pause',
                    'method': 'animate',
                    'args': [[None], {'frame': {'duration': 0, 'redraw': False}, 'mode': 'immediate', 'transition': {'duration': 0}}]
                }
            ]
        }],
        sliders=[{
            'active': 0,
            'yanchor': 'top',
            'xanchor': 'left',
            'currentvalue': {'font': {'size': 13, 'color': '#ddd'}, 'prefix': 'Mission Elapsed Time: ', 'visible': True, 'xanchor': 'right'},
            'transition': {'duration': 80},
            'pad': {'b': 10, 't': 40},
            'len': 0.9,
            'x': 0.05,
            'y': 0,
            'steps': slider_steps
        }],
        scene=dict(
            xaxis=dict(title='Downrange X (m)', range=[0, 12000e3]),
            yaxis=dict(title='Crossrange Y (m)', range=[0, 10000e3]),
            zaxis=dict(title='Altitude Z (m)', range=[0, 1500e3]),
            aspectmode='manual',
            aspectratio=dict(x=2.4, y=1.8, z=0.6),
            camera=dict(eye=dict(x=-1.9, y=-2.1, z=1.2))
        ),
        template='plotly_dark',
        width=1000,
        height=720,
        margin=dict(l=0, r=0, b=0, t=60)
    )
    
    return fig

playable_fig = build_staged_wave_3d_visualizer(flight_log, simulated_targets, n_frames=100, t_total_sim=320.0)
playable_fig.show()
print("Staged Wave 3D Theater Visualizer rendered with Play/Pause controls and inter-wave evaluation pauses.")
```
